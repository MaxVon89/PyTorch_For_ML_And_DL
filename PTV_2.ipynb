## Random TENSORS
    NN use random tensors at first then update iteratively

```python
import torch
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print(torch.__version__)
```

```python
#Create random TENSOR of size (n, m)

randomTENSOR = torch.rand(3,3,4)
randomTENSOR

```

```python
print(f"NDIM: {randomTENSOR.ndim}, Shape: {randomTENSOR.shape}")
```

```python
#Random TENSOR in the shape of an image TENSOR

randomImgTENSOR = torch.rand(size=(224,224,3))
print(randomImgTENSOR)
```

### Zeroes and Ones

```python
# A tensor that only has zeroes (useful for masks)

allZeroes = torch.zeros(3,4)
allZeroes
```

```python
# A tensor that only has ones

allOnes = torch.ones(3,4)
allOnes
```

```python
allZeroes * allOnes
```
