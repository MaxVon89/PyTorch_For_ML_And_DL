### Zeroes and Ones

```python
# A tensor that only has zeroes (useful for masks)

allZeroes = torch.zeros(3,4)
allZeroes
```

```python
# A tensor that only has ones

allOnes = torch.ones(3,4)
allOnes
```

```python
allZeroes * allOnes
```

```python
rT1 = torch.rand(3,4)
print(rT1)


rT2 = rT1 - allOnes
rT2
```

#### Creating a range of TENSORS

```python
# rangeTen = torch.arange(start, end, step)
rangeTen = torch.arange(0, 10)
print(rangeTen)

rgT1 = torch.arange(0,11, 2)
print(rgT1)

```

#### TENSORS like other TENSORS

```python
# tensors like

tenZeroes = torch.zeros_like(rangeTen)
print(tenZeroes)

tenOnes = torch.ones_like(rangeTen)
print(tenOnes)
```

### Tensor Datatypes

```python
tensorFloat32 = torch.tensor([[3.0],[2.0]])
print(f"Shape: {tensorFloat32.shape}, NDIM: {tensorFloat32.ndim}")
print(tensorFloat32.dtype)

tensorFloat16 = torch.tensor([[3.0],[2.0]], dtype=torch.float16)
print(f"Shape: {tensorFloat16.shape}, NDIM: {tensorFloat16.ndim}")
print(tensorFloat16.dtype)
```

```python
# Other important parameters

# torch.tensor([], dtype=torch.float32,
#                  device=cuda, #which device the tensor lives on
#                  requires_grad=False) #whether to track gradience for this tensor
# torch.tensor([], dtype=torch.float32, device=GPU, requires_grad=False)
# torch.tensor([], dtype=torch.float32, device=TPU, requires_grad=False)
```
